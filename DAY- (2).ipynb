Operators


```python
a = 10
b = 5

# 1. Addition
print("1. Addition:", a + b)
```

```python
# 2. Subtraction
print("2. Subtraction:", a - b)
```

```python
# 3. Multiplication
print("3. Multiplication:", a * b)
```

```python
# 4. Division
print("4. Division:", a / b)
```

```python
# 5. Floor Division
print("5. Floor Division:", 17 // 5)
```

```python
# 6. Modulus
print("6. Modulus:", 17 % 5)
```

```python
# 7. Exponent
print("7. Exponent:", 2 ** 3)
```

```python
# 8. Equal
print("8. Equal:", a == b)
```

```python
# 9. Not Equal
print("9. Not Equal:", a != b)
```

```python
# 10. Greater Than
print("10. Greater Than:", a > b)
```

```python
# 11. Less Than
print("11. Less Than:", a < b)
```

```python

# 12. Greater Than or Equal
print("12. Greater Than or Equal:", a >= b)
```

```python
# 13. Less Than or Equal
print("13. Less Than or Equal:", a <= b)
```

```python
# 14. Logical AND
print("14. Logical AND:", a > 5 and b > 2)
```

```python
# 15. Logical OR
print("15. Logical OR:", a > 20 or b > 2)
```

```python
# 16. Logical NOT
print("16. Logical NOT:", not(a > 20))
```
